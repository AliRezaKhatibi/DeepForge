# DeepForge notebook smoke demo
Run in Colab or VS Code with a Python Jupyter kernel. Synthetic CPU-only data.

```python
from pathlib import Path
from tempfile import TemporaryDirectory
from deepforge.monitor import TrainingMonitor
from deepforge.notebook import NotebookMonitor, snapshot_html
```

```python
with TemporaryDirectory() as directory:
    path = Path(directory) / 'live.json'
    monitor = TrainingMonitor(path, epochs=1, batches_per_epoch=3, publish_interval=0)
    view = NotebookMonitor(path, interval=0)
    view.refresh()
    for batch in range(1, 4):
        monitor.batch(epoch=1, batch=batch, metrics={'loss': 1 / batch})
        view.refresh()
    monitor.finish()
    view.refresh(force=True)
    assert monitor.snapshot()['status'] == 'completed'
    assert '100.0%' in snapshot_html(path)
```
